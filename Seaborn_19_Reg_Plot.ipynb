# **Seaborn Library for Data Analytics**

#### **What is Reg Plot :**
##### **Reg Plot = Regression Plot :**
##### **It combines:**
##### Scatter Plot
##### Regression Line (Best Fit Line)
##### **Definition :** A Reg Plot visualizes the relationship between two numerical variables and fits a regression line to show the trend.

| Experience | Salary |
| ---------- | ------ |
| 1          | 3.5    |
| 2          | 4.2    |
| 5          | 6.5    |
| 10         | 10.5   |
| 20         | 20     |
#### **Why Use Reg Plot :**
##### Trend Analysis
##### Correlation Analysis
##### Linear Relationship Detection
##### Machine Learning EDA
##### Predictive Analysis
##### Regression Understanding
##### **Syntax :** sns.regplot(
    data=None,
    x=None,
    y=None,
    scatter=True,
    fit_reg=True,
    ci=95,
    n_boot=1000,
    order=1,
    logistic=False,
    lowess=False,
    robust=False,
    truncate=True,
    x_jitter=None,
    y_jitter=None,
    scatter_kws=None,
    line_kws=None,
    color=None,
    marker='o'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **x :** Independent Variable.(example : x='Experience_Years')
##### **y :** Dependent Variable.(Example : y='Salary_LPA')

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA')
plt.title("Regression Plot")
plt.show()
```

#### **scatter :**
##### Show scatter points.(True/False)

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=True)
plt.title("Regression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=False,)
plt.title("Regression Plot")
plt.show()
```

#### **fit_reg :**
##### Show regression line.*True/False)

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=False,fit_reg=True)
plt.title("Regression Plot")
plt.show()
```

#### **ci :**
##### Confidence Interval.(95/90...)

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=False,fit_reg=True,ci=23)
plt.title("Regression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=False,fit_reg=True,ci=100)
plt.title("Rehression Plot")
plt.show()
```

#### **n_boot :**
##### Bootstrap samples.

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=False,fit_reg=True,n_boot=1000)
plt.title("Regression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter=False,fit_reg=True,n_boot=500)
plt.title("Rehression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA')
plt.title("Regression Plot")
plt.show()
```

#### **order :**
##### Polynomial Regression Degree.

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',order=3)
plt.title("Regression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',order=2)
plt.title("Rehression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',order=1)
plt.title("Rehression Plot")
plt.show()
```

#### **logistic :**
##### Logistic Regression.(True/False)

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',logistic=False)
plt.title("Regression Plot")
plt.show()
```

#### **truncate :**
##### Limit regression line to data range.

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',truncate=True)
plt.title("Regression Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',truncate=False)
plt.title("Rehression Plot")
plt.show()
```

#### **x_jitter :** 
##### Adds random variation on X-axis.
##### **Example :** x_jitter=0.2

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',x_jitter=0.4)
plt.title("Rehression Plot")
plt.show()
```

#### **j_jitter :** 
##### Adds random variation on Y-axis.
##### **Example :** y_jitter=0.2

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',y_jitter=0.6)
plt.title("Regression Plot")
plt.show()
```

#### **color & marker :**
##### Marker Values : 'o'  , 's'  , '^'  , 'D'  , 'X'  , '*'  , '+'

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',color='magenta',marker='^')
plt.title("Regression Plot")
plt.show()
```

#### **scatter_kws :** Customize scatter points.
##### **Syntax :** scatter_kws={'s':100,'alpha':0.7,'color':'red','marker':'o'}

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter_kws={'s':100,'alpha':0.7,'color':'red','marker':'o'})
plt.title("Regression Plot")
plt.show()
```

#### **line_kws :**
##### Customize regression line.
##### **Example :** line_kws={'linewidth':3,'linestyle':'--','color':'blue'}

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df,x='Experience_Years',y='Salary_LPA',scatter_kws={'s':100,'alpha':0.7,'color':'red','marker':'o'},line_kws={'linewidth':3,'linestyle':'--','color':'blue'})
plt.title("Regression Plot")
plt.show()
```

| Scatter Plot       | Reg Plot                   |
| ------------------ | -------------------------- |
| Only points        | Points + Regression Line   |
| Shows relationship | Shows relationship + trend |
