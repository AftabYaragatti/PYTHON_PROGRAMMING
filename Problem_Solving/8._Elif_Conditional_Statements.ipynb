## 3. Elif Conditional Statements

#### LEVEL 1 – BASIC ELIF

#### 1. Check whether a number is positive, negative, or zero.

```python
num=int(input())
if num>0:
    print('Positive')
elif num==0:
    print('Zero')
else:
    print('Negative')
```

#### 2. Grade system:
####   * Marks ≥ 90 → A
####   * ≥ 75 → B
####   * ≥ 50 → C
####   * Else → Fail

```python
marks=int(input())
if marks>=90:
    print('Grade A')
elif marks>=75:
    print('Grade B')
elif marks>=50:
    print('Grade C')
else:
    print('Fail')
```

#### 3. Check whether a number is 1-digit, 2-digit, or 3-digit.

```python
num=int(input())
if len(str(num))==1:
    print('1 Digit Number')
elif len(str(num))==2:
    print('2 Digit Number')
elif len(str(num))==3:
    print('3 Digit Number')
else:
    print('More than 3 Digit Number')
```

#### 4. Check traffic signal (Red / Yellow / Green).

#### 5. Check day type (Weekday / Weekend).

```python
day=input()
if day in ['Monday','Tuesday','Wednesday','Thursday','Friday']:
    print('Weekday')
elif day=='Saturday':
    print('Weekend')
else:
    print('Sunday')
```

#### 6. Check month season (Winter / Summer / Rainy).

```python
month=int(input())
if (month>=1 and month<=4):
    print('Summer Season')
elif (month>=5 and month<=8):
    print('Rainy Season')
else:
    print('Winter Season')
```

#### 7. Check whether age is child, teen, adult, or senior.

```python
age=int(input())
if (age>=1 and age<=12):
    print('Child')
elif (age>=13 and age<=20):
    print('Teen')
elif (age>=21 and age<=50):
    print('Adult')
else:
    print('Senior')
```

#### 10. Simple calculator using +, -, *, /.

```python
num1=int(input())
num2=int(input())
operator=input()
if operator=='+':
    print(num1+num2)
elif operator=='-':
    print(num1-num2)
elif operator=='*':
    print(num1*num2)
elif operator=='/':
    print(num1/num2)
```

### LEVEL 2 – INTERMEDIATE

#### 11. Electricity bill calculation (slab system).

```python
units=int(input())
if units<=100:
    print(units*1.5)
elif units<=300:
    print(units*1.5+(units-100)*2.5)
elif units<=500:
    print(units*1.5+(units-100)*2.5+(units-300)*3.5)
```

#### 12. Income tax calculation (slab-based).

```python
income=int(input())
if income<=250000:
    print('Zero Income Tax')
elif income<=5000000:
    print((income-250000)*0.05)
elif income<=10000000:
    print((income-250000)*0.5+(income-500000)*0.1)
```
