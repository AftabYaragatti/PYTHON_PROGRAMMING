#### 31. Sort three numbers using only comparison operators.

```python
num1=int(input())
num2=int(input())
num3=int(input())
if num1>num2:
    num1,num2=num2,num1
if num2>num3:
    num2,num3=num3,num2
if num1>num2:
    num1,num2=num2,num1
print(num1,num2,num3)
```

#### 32. Check whether a number is palindrome (compare original and reversed).

```python
num=int(input())
rev=(((num//100)*1)+(((num//10)%10)*10)+((num%10)*100))
res=(num==rev)
print(res)
```

#### 33. Validate strong password (length >= 8).

```python
EP=input()
SP=((EP[0]>='A' and EP[0]<='Z') and (EP[1:3]>='a' and EP[1:3]<='z') 
    and ( not( (EP[4]>='A' and EP[4]!='Z') and (EP[4]>='a' and EP[4]!='z') and (EP[4]>='0' and EP[4]>='9') ) ) and (EP[5:9]>='0' and EP[5:9]<='9'))
print(SP,EP)
```

#### 37. Check if a number is Armstrong (compare sum with original).

```python
num=int(input())
s=str(num)
digit=len(s)
rev=(int(s[0])**digit+int(s[1])**digit+int(s[2])**digit)
print(num==rev)
```
