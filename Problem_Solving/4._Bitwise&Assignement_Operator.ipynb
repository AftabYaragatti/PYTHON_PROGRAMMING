#### 11. Check whether a number is even using bitwise.

```python
num=int(input())
res=(num & 1 ==0) #6-110 #1-001 => 110+001=000 
print(res)
```

#### 12. Multiply a number by 2 using left shift.

```python
num=int(input()) 
res=(num<<1) #5<<2 -> 5*2=10
print(res*2) #10*2=20
# input-5
# 5-101
# left shift 101-1010 (0 adds at the end)
# 1010 in decimal -> 10
# 5*2 =10, 10*2=20
```

#### 13. Divide a number by 2 using right shift.

```python
num=int(input()) 
res=(num>>1) #5>>2 -> 5//2=2
print(res//2) #2//2=1
# input-5
# 5-101
# left shift 101-01 (removes last bit)
# 01 in decimal -> 1
# 5//2 =4 , 4//2=1
```

#### 14. Swap two numbers using XOR.

```python
num1=int(input())
num2=int(input())
print('Before :',num1,num2)
num1=num1^num2
num2=num1^num2
num1=num1^num2
print('After :',num1,num2)

# 4 - 100
# 5 - 101
# num1 = 100 ^ 101 = 110  (same bits - 1, else different bits 0)
# num2 = 110 ^ 101 = 100
# num1 = 110 ^ 100 = 101

# 000
# 001
# 010
# 011
# 100
# 101
# 110
# 111
```

#### 15. Check if a number is power of 2 using bitwise.

```python
num=int(input())
res=(num & (num-1))==0
print(res)
# 4 - 100
# num-1 = 4 - 1 = 3 => 011
# 4 & 3 = 100 & 011 = 000 
# 000 - 0 == 0
```

#### 16. Count number of set bits.

```python
num=int(input())
count=0
while num:
    num=(num & (num-1))
    count+=1
print(count)
# 4 - 100 (1 Set Bits) # only 1 bit is counting
# 7 - 111 (3 Set Bits)
```

#### 17. Check if specific bit is set or not.

```python
num=int(input())
k=int(input())
num=((num & (1<<k))!=0)
print(num)
# 5 - 101 , k = 2 = 010
# num => 5 & (1<<2) = 101 & (1*2=2,2*2=4) = 101 & 100 = 100 (4!=0)
```

### 5. ASSIGNMENT OPERATOR

####  | Operator | Meaning                 | Example   | Same As      |
##### | `+=`     | Add and assign          | `x += 5`  | `x = x + 5`  |
##### | `-=`     | Subtract and assign     | `x -= 3`  | `x = x - 3`  |
##### | `*=`     | Multiply and assign     | `x *= 2`  | `x = x * 2`  |
##### | `/=`     | Divide and assign       | `x /= 4`  | `x = x / 4`  |
##### | `//=`    | Floor divide and assign | `x //= 2` | `x = x // 2` |
##### | `%=`     | Modulus and assign      | `x %= 3`  | `x = x % 3`  |
##### | `**=`    | Power and assign        | `x **= 2` | `x = x ** 2` |

#### BASIC
##### 1. Add 10 to a variable using `+=`
##### 2. Multiply variable by 10 using `*=`
##### 3. Substract 10 to a variable using `-=`
##### 4. Divide number by 10 using `/=`
##### 5. Modulas number by 10 using `%=`
##### 6. Floor Division number by 10 using `//=`
##### 7. Power number by 10 using `/=`

```python
num=int(input())
num+=10
print('Add :',num)
num-=10
print('Sub :',num)
num*=10
print('Mul :',num)
num/=10
print('Div :',num)
num//=10
print('Floor Div :',num)
num%=10
print('Modulas :',num)
num**=10
print('Power :',num)
```
