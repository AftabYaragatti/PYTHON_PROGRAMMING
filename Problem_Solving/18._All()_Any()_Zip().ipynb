## ALL(),ANY(),ZIP() FUNCTIONS

### `any()` Practice Questions

### Basic

#### 1. Check if a list contains at least one even number.

#### 2. Check if a string contains at least one digit.

#### 3. Check if a list has any negative numbers.

#### 4. Check if at least one word in a list starts with capital letter.

#### 5. Check if any element in list is divisible by 5.

### Intermediate

#### 6. Check if any student scored above 90.

#### 7. Check if any word length is greater than 10.

#### 8. Check if any value in dictionary is zero.

#### 9. Check if any row in a 2D list contains negative numbers.

#### 10. Check if any file name ends with `.txt`.

### Advanced

#### 11. Check if any sublist contains duplicate elements.

#### 12. Check if any two numbers in list sum to target.

#### 13. Check if any word is palindrome in a list.

#### 14. Check if any element appears more than once.

#### 15. Validate if any password violates rules.

### all() Practice Questions

### Basic

#### 1. Check if all numbers are positive.

#### 2. Check if all elements are even.


#### 3. Check if all characters in string are lowercase.


#### 4. Check if all values in list are integers.


#### 5. Check if all numbers are greater than 10.

### Intermediate


#### 6. Check if all students passed (marks > 40).


#### 7. Check if all words contain vowel.


#### 8. Check if all dictionary keys are strings.


#### 9. Check if all rows in 2D list have equal length.


#### 10. Check if all file names end with `.py`.


### Advanced


#### 11. Validate if all passwords meet strength criteria.


#### 12. Check if all elements are unique.


#### 13. Check if all numbers in list are prime.


#### 14. Check if all sublists are sorted.


#### 15. Validate if all required fields exist in dictionary.
