#### 1. Extract Only Vowels or Digits or Both

##### Q1: Given the string 'hello world 123 haii', extract only the vowels and digits in order.
##### A1: 'eoo123aii'

```python
# Using While Loop :
st='hello world 123 haii'
i=0
res=''
while i<len(st):
    ch=st[i]
    if ch in 'aeiouAEIOU0123456789':
        res=res+ch
    i+=1
print(res)
```

##### Q12: Given the string 'HELLOWORLD', convert to lowercase and replace vowels with * , e with $ :
##### A12: 'h\$ll\*w\*rld '

```python
# Using While Loop :
st='HELLOWORLD'
i=0
res=''
while i<len(st):
    ch=st[i]
    if ch>='A' and ch<='Z':
        res+=chr(ord(ch)+32)
        if ch in 'aeiouAEIOU':
            if 'O'==ch:
                res=res+ch.replace('O','*')
            if 'E'==ch:
                res=res+ch.replace('E','$')
    i+=1
print(res)
```

#### 9. Word Indexing / Labeling

##### Q28: Given 'HAII HELLO', append word-wise index to each word’s characters:
##### A28: 'HA1I2I3 HE4L5L6O7'

```python
# Using While Loop :
st="HAII HELLO"
words=st.split()
res=""
count=1
w=0
while w<len(words):
    word=words[w]
    i=0
    while i<len(word):
        if i<1:   # First letter is unchanged
            res+=word[i]
        else:
            res+=word[i]+str(count)
            count+=1
        i+=1
    if w!=len(words)-1:
        res+=" "
    w+=1
print(res)
```

```python
# Using For Loop :
st="HAII HELLO"
words=st.split()
res=""
count=1
w=0
for w in range(len(words)):
    word=words[w]
    for i in range(len(word)):
        if i<1:   # First letter is unchanged
            res+=word[i]
        else:
            res+=word[i]+str(count)
            count+=1
    if w!=len(words)-1:
        res+=" "
print(res)
```
