#### 7. WAP to check whether sum of digits and product of the digits in a given number is equal or not.If it is equals then print **SPY NUMBER** Else Not a SP NUMBER

### **SPY NUMBER PROGRAM**
#### Take any digits of a number, the sum of the digits are equals to the products of digits
#### **sum** of digits == **product** of digits
#### Example : 123
#### 1+2+3==1*2*3
#### 6==6

```python
num=int(input())
sum=0
prod=1
while num!=0:
    ld=num%10
    sum+=ld
    prod*=ld
    num=num//10
if sum==prod:
    print("SPY NUMBER")
else:
    print("NOT A SPY NUMBER")
```

```python
num=int(input())
square=num**2
sum=0
while square!=0:
    ld=square%10
    sum+=ld
    square//=10
if sum==num:
    print("NEON Number")
else:
    print("Not A NEON Number")
```

#### 11. WAP to print the final sum of the factorial of each digit in a given number (Fact - Product of the series from 1 to n)

### **STRONG NUMBER PROGRAM**
#### Take any number of the digits that should be the **sum** of the **factorial** of the number is equals to the original number
#### **Sum** of the **factorial(Product)** of the number == **Original** Number
#### Example : num=145
#### fact of each digit : 5 - 1*2*3*4*5=120, 4 - 1*2*3*4=24, 1 - 1 = 1
#### sum of fact : 120+24+1=145
#### Original Number == **Sum** of the **factorial(Product)** of the number
#### 145 == 145

```python
num=int(input())
num1=num
sum=0
while num!=0:
    ld=num%10
    start=1
    prod=1
    while start<=ld:
        prod*=start
        start+=1
    num//=10
    sum+=prod
    print(f'For digit {ld}, The Product of series starts from 1 to {ld} : {prod}')
print(f'Sum of all Product is {sum}')
if num1==sum:
    print('Strong Number')
else:
    print('Not a Strong Number')
```

```python
num=int(input())
count=0
num1=num
while num!=0:
    count+=1
    num//=10
print('Count :',count)
Sum=0
while num1!=0:
    ld=num1%10
    num1//=10
    Sum=Sum+(ld**count)
print(Sum)
if num1==Sum:
    print('Armstrong Number')
else:
    print('Not a Armstrong Number')
```
