# _**1. PRIME NUMBER PROGRAM :**_
### The given number should be only divisible by 1 and that number itself
### Example : 
### number=7
### 7%1==0 True - count=1
### 7%2==0 False
### 7%3==0 False
### 7%4==0 False
### 7%5==0 False
### 7%6==0 False
### 7%7==0 True - count=1+1=2
### count==2
### 2==2

### **1.1 CHECK NUMBER IS PRIME NUMBER OR NOT**

```python
num=int(input())
start=1
count=0
while start<=num:
    if num%start==0:
        count+=1
    start+=1
if count==2:
    print('Prime Number')
else:
    print('Not a Prime Number')
```

### **1.2 PRINT THE PRIME NUMBERS FROM M TO N**

```python
m=int(input('Enter the Start Number :'))
n=int(input('Enter the End Number :'))
print(f'Prime Numbers from {m} to {n} is :')
while m<=n:
    num=m
    start=1
    count=0
    while start<=num:
        if num%start==0:
            count+=1
        start+=1
    if count==2:
        print(num,end=' | ')
    m+=1
```

### **1.3 PRINT THE FIRST N PRIME NUMBERS**

```python
num=1
c=0
n=int(input('Enter the First N Number :'))
print(f'First {n} Prime Numbers is :')
while c<n:
    num1=num
    start=1
    count=0
    while start<=num1:
        if num1%start==0:
            count+=1
        start+=1
    if count==2:
        print(num1,end=' | ')
        c+=1
    num+=1
```

# **2. FACTORIAL NUMBER PROGRAM :**
### The given number should be product of the number from that number to till it reaches to 1 ( n * (n-1) )
### Example :
### number=5 
### 5\*(5-1) = 5\*4 = 20
### 5\*(5-1)\*(4-1) = 20\*3 = 60

### **2.1 PRINT THE FACTORIAL OF THE NUMBER**

```python
num=int(input())
fact=1
start=1
while start<=num:
    fact*=start
    start+=1
print(fact)
```

### Sum of digits of the number == Sum of digits of its prime factors
### Note : Number Must not be a Prime , we uses prime factors
### Example : num = 666
### Sum of Digits of a given number
### 6+6+6=18
### Find Prime Factorization
### strat=2 always because we are checking prime factors
### INCOMPLETE #***************************************************************************************************************#
### num2=666, start=2, num2%start, 666%2==0 True , num1=start, num1=2 , sum of that num1 is sum2 = 2 , num2 = 666//2==333 start+=1 -> start = 2+1 = 3
### num2=333, 333%3==0 True , num1=3 , sum of that num1 is sum2 = 3 , num2 = 333//3==111 start+=1 -> start = 3+1=4
### num2=111, 111%4==0 False , num1=4 , sum of that num1 is sum2 = 4 , num2 = 111//3==37 start+=1 -> start = 4+1=5
### num2=37, 37%5==0 False , start=5+1=6
### num2=37, 37%6==0 False , start=6+1=7
### num2=37, 37%7==0 False , start=7+1=8
### ....................................
### num2=37, 37%37==0 True , num1=37 , sum of that num1 is sum2 = 226 , num2 = 37//37==1 start+=1 -> start = 37+1=38

```python
num=int(input())
num1=num
sum=0
# Find the sum of the digits :
while num1!=0:
    d=num1%10
    sum+=d
    num1//=10
# Find Prime Factorization :
sum2=0
num2=num
start=2
while start<=num2:
    while num2%start==0:
        num1=start
        while num1!=0:
            d2=num1%10
            sum2+=d2
            num1//=10
        num2//=start
    start+=1
if sum==sum2:
    print('Smith Number')
else:
    print('Not a Smith Number')
```
