#### 22. Find largest among three numbers using and.

```python
num1=int(input())
num2=int(input())
num3=int(input())
res1=(num1>num2 and num1>num3)
res2=(num2>num1 and num2>num3)
res3=(num3>num2 and num3>num1)
print(res1,res2,res3)
```

#### 23. Validate strong password (length >=8 and contains uppercase and digit).

#### 24. Check if triangle is valid (sum of two sides > third side).

```python
side1=int(input())
side2=int(input())
side3=int(input())
res=((side1+side2>side3) and (side1+side3>side2) and (side3+side2>side1))
print(res)
```
