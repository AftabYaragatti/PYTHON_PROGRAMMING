#### 21. Merging two string character by character 
#### I/P : S1:abc, S2:xyz
#### O/P : axbycz
#### I/P : S1:abcde, S2:xyz
#### O/P : axbyczde

````python
st1=input()
st2=input()
res=''
i=0
while i<len(st1) or i<len(st2):
    if i<len(st1) and i<len(st2):
        res+=st1[i]+st2[i]
    elif i<len(st1):
        res+=st1[i]
    else:
        res+=st2[i]
    i+=1
print(res)
````

````python
st1=input()
st2=input()
res=''
i=0
j=0
while i<len(st1) or j<len(st2):
    if i<len(st1) and j<len(st2):
        if st1[i]<st2[j]:
            res+=st1[i]
            i+=1
        else:
            res+=st2[j]
            j+=1
    elif i<len(st1):
        res+=st1[i]
        i+=1
    else:
        res+=st2[j]
        j+=1
print(res)
````
