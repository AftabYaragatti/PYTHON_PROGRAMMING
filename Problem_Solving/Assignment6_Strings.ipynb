#### 2. WAPT to print
#### Only Alphabets
#### Only Digits
#### Only Special Characters

```python
st=input()
alph=''
dgts=''
schar=''
for i in st:
    if (i>='A' and i<='Z') or (i>='a' and i<='z'):
        alph+=i
    elif (i>='0' and i<='9'):
        dgts+=i
    else:
        schar+=i
print(f'\nAlphabets : {alph}\n Digits : {dgts}\n Special Char : {schar}')
```

#### 4. WAPT extract all the duplicate characters from a given string.

```python
st=input()
res='' #aabcda
res1=''
for i in st:
    if i not in res:
        res+=i
    else:
        res1+=i
print(res1)
```

#### 7. Check if a string is a palindrome.

```python
st=input()
res=''
for i in st:
    res=i+res
if res==st:
    print('String is Palindrome')
else:
    print('Not a Palindrome')
```

#### 8. Count the frequency of each character in a string.

```python
st=input()
res=''
i=0
j=0
while i<len(st):
    count=0
    ch=st[i]
    while j<len(st):
        ch1=st[j]
        if ch==ch1:
            res+=ch1
            count+=1
        else:
            count=1
        j+=1
    i+=1
    print(res,count)
    res=''
```
