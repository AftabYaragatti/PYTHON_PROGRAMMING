```python
# 12 : Write an expression to get the middle digit of a 3-digit number.
n=int(input())
res=(n//10)%10
print(res)
```

```python
# 13 : Write an expression to get the sum of digits of a 2-digit number.
n=int(input())
Sum=0
while n!=0:
    digit=n%10
    Sum+=digit
    n=n//10
print(Sum)
```

```python
# 14 : Write an expression to reverse a 2-digit number.
n=int(input())
rev=(n%10)*10+(n//10)*1
print(rev)
```

```python
# 15 : Write an expression to get the tens digit of a number.
n=int(input())
tens_digit=(n//10)%10
print(tens_digit)
```

```python
# 16 : Write an expression to count how many hundreds are in n.
n=int(input())
res=n//100
print(res)
```

```python
# 17 : Write an expression to check if last digit of n is 7.
n=int(input())
digit=n%10
if digit==7:
    print(digit)
```

```python
# 18 : Write an expression to extract the third digit from right.
n=int(input())
res=n%1000
TD=res//100
print(TD)
```

```python
# 19 : Write an expression to check if a number ends with 0.
n=int(input())
digit=n%10
if digit==0:
    print(n)
```

```python
# 20 : Write an expression to remove last 3 digits of a number.
n=int(input())
res=n//1000
print(res)
```

#### Advanced `%` and `//` Logic

21. Write an expression to check if number is 4-digit.
22. Write an expression to get the first digit of any number.
23. Write an expression to check if last two digits form a number divisible by 4.
24. Write an expression to rotate last digit to front (2-digit number).
25. Write an expression to check if a number is palindrome (2-digit).
26. Write an expression to check if sum of digits is even (2-digit).
27. Write an expression to split a number into quotient and remainder.
28. Write an expression to extract every alternate digit (conceptually).
29. Write an expression to find number of complete dozens in `n`.
30. Write an expression to find remaining items after forming dozens.

```python
# 21 : Write an expression to check if number is 4-digit.
n=int(input())
res=n%10000
if n==res:
    print('4 Digit Number')
```

```python
# 22 : Write an expression to get the first digit of any number.
n=int(input())
digit=n//10**(len(str(n))-1)
print(digit)
```

```python
# 23 : Write an expression to check if last two digits form a number divisible by 4.
n=int(input())
digit=n%100
if digit%4==0:
    print('Last 2 Digit divisible by 4')
```

```python
# 24 : Write an expression to rotate last digit to front (2-digit number).
n=int(input())
Ldigit=n%10 #last digit
Fdigit=n//10 # First Digit
print('Orginal :',n)
print('After Rotate :',str(Ldigit)+str(Fdigit))
```

```python
# 25 : Write an expression to check if a number is palindrome (2-digit).
n=int(input())
res=(n%10)*10+(n//10)*1
if n==res:
    print('Palindrome',res)
```

```python
# 26 : Write an expression to check if sum of digits is even (2-digit).
n=int(input())
Sum=0
while n!=0:
    digit=n%10
    Sum+=digit
    n=n//10
print(Sum)
```

```python
# 27 : Write an expression to split a number into quotient and remainder.
n=int(input())
quotient=n//10
reminder=n%10
if str(quotient) in str(n):
    print('Quotient :',quotient)
if str(reminder) in str(n):
    print('Reminder :',reminder)
```

```python
# 28 : Write an expression to extract every alternate digit (conceptually).
n=int(input())
while n!=0:
    digit=n%10
    print(digit)
    n=(n//100)
```

```python
# 29 : Write an expression to find number of complete dozens in n.
n=int(input())
dozens=n//12
print(dozens)
```

```python
# 30 : Write an expression to find remaining items after forming dozens.
n=int(input())
dozens=n//12
rem=n%12
print('Dozens :',dozens)
print('Remaining :',rem)
```
