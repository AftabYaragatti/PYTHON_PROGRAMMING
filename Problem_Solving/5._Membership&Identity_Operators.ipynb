## 6. MEMBERSHIP OPERATORS 

### LEVEL 1 – BASIC

#### 1. Check if 5 is in list [1,2,3,4,5].

```python
l=list(map(int,input().split()))
n=int(input())
if n in l:
    print(f'{n} Present in List')
else:
    print(f'{n} Not Present in List')
```

#### 2. Check if "a" is in "apple".

```python
ch=input()
st='apple'
if ch in st:
    print(f'{ch} Present in String')
else:
    print(f'{ch} Not Present in String')
```

#### 3. Check if 10 is not in tuple (1,2,3).

```python
t=tuple(map(int,input().split()))
num=int(input())
if num in t:
    print(f'{t} Present in Tuple')
else:
    print(f'{t} Not Present in Tuple')
```

#### 4. Check if "x" not in "python".

```python
s='python'
st=input()
if st not in s:
    print(f'{st} Not Present in String')
else:
    print(f'{st} Present in String')
```

#### 5. Check if element exists in set.

```python
s=set(map(int,input().split()))
ele=int(input())
if ele in s:
    print(f'{ele} Present in Set')
else:
    print(f'{ele} Not Present in Set')
```

#### 6. Check if key exists in dictionary.

```python
d=dict({input():input(),input():input(),input():input()})
key=input('Enter Key :')
if key in d:
    print(f'{key} Present in Dictionary')
else:
    print(f'{Key} Not Present in Dictionary')
```

#### 7. Check if value exists in dictionary.

```python
d=dict({input():input(),input():input(),input():input()})
value=input('Enter Value :')
if value in d.values():
    print(f'{value} Present in Dictionary')
else:
    print(f'{value} Not Present in Dictionary')
```

#### 8. Check if number in range(1,10).

```python
num=int(input())
if num in range(1,10):
    print(f'{num} Present in Range')
else:
    print(f'{num} Not Present in Range')
```

#### 9. Check if character is vowel using in.

```python
ch=input()
if ch in 'aeiouAEIOU':
    print(f'{ch} Present in Vowel;')
elif (ch>='A' and ch<='Z') or (ch>='a' and ch<='z'):
    print(f'{ch} is Consonant')
else:
    print(f'{ch} Not an Alphabet')
```

#### 10. Check if word is in sentence.

```python
sentence=input()
word=input()
if word in sentence:
    print(f'{word} Present in Sentence')
else:
    print(f'{word} Not Present in Sentence')
```

### LEVEL 2 – INTERMEDIATE

#### 11. Validate username exists in list.

```python
username=input()
user_Name_list=['Aftab','Basavaraj','Anil','Anand','Nikhil','Bhagya','Soudamini','Spoorthi','Swati']
if username in user_Name_list:
    print(f'{username} Present in List')
else:
    print(f'{username} Not Present in List')
```

#### 12. Check if substring exists in string.

```python
string=input()
sub_str=input()
if sub_str in string:
    print(f'{sub_str} Present in String')
else:
    print(f'{sub_str} Not Present in String')
```

#### 13. Count occurrences using membership logic.

```python
lst=list(map(int,input().split()))
ele=int(input())
count=0
for x in lst:
    if x==ele:
        count+=1
print(count)
```

#### 14. Filter list using in.

```python
lst1=list(map(int,input().split()))
lst2=list(map(int,input().split()))
filtered=[x for x in lst1 if x not in lst2]
print(filtered)
```

#### 15. Check duplicate element.

```python
lst=list(map(int,input().split()))
if len(lst)!=len(set(lst)):
    print(set(lst))
```

#### 16. Remove elements not in allowed list.

```python
lst=list(map(int,input().split()))
allowed=[2,4,6,8]
filtered=list(filter(lambda x:x in allowed, lst))
print(filtered)
```

#### 17. Validate email contains "@" and ".com".

```python
email=input()
if '@' and '.com' in email:
    print(email)
```

#### 18. Check if password contains special character.

```python
password=input()
char='@'
if char in password:
    print(f'{password} contains special characters')
```

#### 19. Validate input belongs to allowed choices.

```python
ip=input()
allowed=['A','B','C','D']
if ip in allowed:
    print(ip)
```

#### 20. Check if any digit exists in string.

```python
st=input()
digits='0123456789'
a=any(c in digits for c in st)
print(a)
```

## 7. IDENTITY OPERATORS

### LEVEL 1 – BASIC

#### 1. Compare two integers using is.

```python
a=int(input())
b=int(input())
res=a is b
print(res)
```

#### 2. Compare two integers using ==.

```python
a=int(input())
b=int(input())
res=a==b
print(res)
```

#### 3. Compare two strings using is.

```python
a=input()
b=input()
res=(a is b) 
print(res)
```

#### 4. Compare two lists using is.

```python
l=[1,2,3,4]
l1=[1,2,3,4]
l2=l
print(l is l1)
print(l is l2)
# Check two varibles points to the same object in memory.
```

#### 5. Assign b = a and check a is b.

```python
a=19
b=a
print(a is b)
```

#### 6. Check whether variable is None using is.

```python
a=100
b=None
print(a is None)
print(b is None)
```

#### 7. Check whether two variables refer to same object.

```python
var1=100
var2=200
var3=100
var4=var2
print(var1 is var2)
print(var1 is var3)
print(var2 is var4)
```

#### 8. Compare two tuples using is.

```python
t1=1,2,3
t2=1,2,3
print(t1 is t2) #False Because two variables are pointing to different objects
t3=t1
print(t1 is t3)
```

#### 9. Check if two boolean values are identical.

```python
a=True
b=True
print(a is b)
```

#### 10. Compare float variables using is.

```python
a=10.2
b=10.2
print(a is b)
```
