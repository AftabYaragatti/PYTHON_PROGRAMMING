#### 1. Check if a number is positive, and if positive check whether it is even or odd.

```python
num=int(input())
if num>0:
    if num%2==0:
        print('Even')
    else:
        print('Odd')
```

#### 10. Check if number is 3-digit, then check if palindrome.

```python
num=int(input())
if len(str(num))==3:
    if str(num)==str(num)[::-1]:
        print(num)
```

### LEVEL 2 – INTERMEDIATE

#### 11. Find largest among three numbers using nested if.

```python
num1=int(input())
num2=int(input())
num3=int(input())
if num1>num2:
    if num1>num3:
        print(num1)
elif num2>num3:
    print(num2)
else:
    print(num3)
```

#### 12. Check if year is leap year using nested conditions.

```python
year=int(input())
if year%4==0:
    if year%100==0:
        if year%400==0:
            print('Leap Year')
        else:
            print('Not a Leap Year')
    else:
        print('Leap Year')
```

#### 16. Triangle validation:
#### * Check positive sides
#### * Then check triangle condition

```python
side1=int(input())
side2=int(input())
side3=int(input())
if side1+side2>side3:
    if side1+side3>side2:
        if side2+side3>side1:
            print(side1,side2,side3)
```

#### 17. Check if number is Armstrong (nested logic).

```python
num=int(input())
digit=len(str(num))
if int(num):
    res=(num//100)**digit+((num//10)%10)**digit+(num%10)**digit
    if num==res:
        print(res)
```

#### 18. Prime number check using nested loop + if.

```python
num=int(input())
count=0
start=1
while num>=start:
    if num%start==0:
        count+=1
    start+=1
if count==2:
    print('Prime Number')
else:
    print('Not a Prime Number')
```

#### 19. Profit or loss calculation.

```python
price=int(input())
sale=int(input())
if price<sale:
    if price!=sale:
        print('Profit')
elif price>sale:
    print('Loss')
else:
    print('No Profit,No Loss')
```
