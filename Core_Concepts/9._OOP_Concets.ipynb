#### 3.Destructor Methods :

```python
# A destructor is a special method that is automatically called when an object is destroyed (deleted from memory).
class Employee:
    def __init__(self,ename,esal):
        self.ename=ename
        self.esal=esal
    def show(self):
        print(f'Employee Name : {self.ename}\nEmployee Salary : {self.esal}')
        # After initialization i can modify the constructor
        self.ename='Smith'
        self.esal=91000
        print(f'Employee Name : {self.ename}\nEmployee Salary : {self.esal}')
    def __del__(self):
        print("Destructor called")
e1=Employee('Aftab',98000)
e1.show()
del e1
```

### Types of methods:

#### 1.Instance Methods (Object Methods):

```python
# It is a method which is used to access and modify an objects and for all methods passing self is manadatory.
class Employee:
    def __init__(self,ename,esal):
        self.ename=ename #### Instance Varible
        self.esal=esal #### Instance Varible
    def details(self): #Instance Method
        self.ename='Smith' #Instance Methods modify
        print(f'Employee name : {self.ename}\nSalary : {self.esal}')
emp=Employee('Aftab',98000)
emp.details() # call instance method
```

#### 2.Class Methods(cls):

```python
# It is a method which is used to access and modify an members of class and for all the method we use decorator @classmethod is manadatory
# Instead of an passing an self pass the cls is manadatory.
class Employee:
    cname="TCS" #class variable
    def __init__(self,ename,esal):
        self.ename=ename 
        self.esal=esal 
        print(f'Employee name : {self.ename}\nSalary : {self.esal}')
    @classmethod
    def details(cls,n): #Class Method
        print(f'Company Name : {Employee.cname}')
        # modify class methods using cls
        cls.cname=n
        print(f'Company Name : {cls.cname}')
Employee.details('Infosys') # call instance method
emp=Employee('Aftab',98000)
```

#### 3. Static Method :

```python
# It is a method which is neither belongs to the class or an object but it works as useful method for class and object.No  need to create an object and no need to pass self or cls . @staticmethod is manadatory
# Instead of an passing an self pass the cls is manadatory. It works like a normal method
class Employee:
    @staticmethod
    def details(name,sal): #Static Method
        print(f'Name : {name}\nSalary : {sal}')
Employee.details('Aftab',98000) # call static method
```

### Constructor Chining in Single Level Inheritance

```python
# It is a process where one constructor calls another constructor either from same class or from parent class ,To ensure parent class constructor execute before child class
class Parent:
    def __init__(self,pname,page):
        self.pname=pname
        self.page=page
class Child(Parent):
    def __init__(self,pname,page,cname,cage):
        super().__init__(pname,page)
        self.cname=cname
        self.cage=cage
    def display(self):
        print(f'Parent Name : {self.pname}\nParent Age : {self.page}\nChild Name : {self.cname}\nChild Age : {self.cage}')
c=Child("Smith",41,"Jhon",14)
c.display()
```

#### 2.Multilevel Inheritance :

```python
# It is a process where one constructor calls another constructor either from same class or from parent class ,To ensure parent class constructor execute before child class
class GParent:
    def __init__(self,gname,gage):
        self.gname=gname
        self.gage=gage
    def displayG(self):
        print(f'Grand Parent Name : {self.gname}\nGrand Parent Age : {self.gage}')
        return self
class Parent(GParent):
    def __init__(self,gname,gage,pname,page):
        super().__init__(gname,gage)
        self.pname=pname
        self.page=page
    def displayP(self):
        print(f'Parent Name : {self.pname}\nParent Age : {self.page}')
        return self
class Child(Parent):
    def __init__(self,gname,gage,pname,page,cname,cage):
        super().__init__(gname,gage,pname,page)
        self.cname=cname
        self.cage=cage
    def displayC(self):
        print(f'Child Name : {self.cname}\nChild Age : {self.cage}')
        return self
c=Child("Smith",78,"Jhon",44,"Jafri",14)
c.displayG().displayP().displayC()
```

#### 4.Hierarchical Inheritance :

```python
# It is a process where one constructor calls another constructor either from same class or from parent class ,To ensure parent class constructor execute before child class
class GParent:
    def displayG(self):
        self.pname="David"
        self.page=74
        print(f'Grand Parent Name : {self.pname}\nGrand Parent Age : {self.page}')
class Parent1(GParent):
    def displayP1(self):
        self.pname="Smith"
        self.page=34
        self.displayG()
        print(f'Parent Name : {self.pname}\nParent Age : {self.page}')
class Parent2(GParent):
    def displayP2(self):
        self.pname="Jhon"
        self.page=37
        self.displayG()
        print(f'Parent Name : {self.pname}\nParent Age : {self.page}') 
p1=Parent1()
p1.displayP1() # Same variables hence parent variable is accessible instead of child variable
p2=Parent2()
p2.displayP2()
```
