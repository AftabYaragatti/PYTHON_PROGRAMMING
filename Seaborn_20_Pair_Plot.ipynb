# **Seaborn Library for Data Analytics**

#### **What is Pair Plot :**
##### A Pair Plot creates scatter plots between all numerical columns and histograms/KDE plots on the diagonal.
##### **Definition :** Pair Plot visualizes relationships among multiple numerical variables in a dataset in a single figure.
##### **Why Use Pair Plot :**
##### Feature Relationship Analysis
##### Correlation Detection
##### Outlier Detection
##### Cluster Detection
##### Exploratory Data Analysis (EDA)
##### Machine Learning Feature Understanding
##### **Example :** For columns:
##### Age , Experience , Salary , Performance
##### It automatically creates:
##### Age vs Experience
##### Age vs Salary
##### Age vs Performance
##### Experience vs Salary
##### Experience vs Performance
##### Salary vs Performance
##### Plus histograms/KDE plots on the diagonal.

##### **Syntax :** sns.pairplot(
    data,
    hue=None,
    vars=None,
    x_vars=None,
    y_vars=None,
    kind='scatter',
    diag_kind='auto',
    markers=None,
    palette=None,
    corner=False,
    height=2.5,
    aspect=1,
    plot_kws=None,
    diag_kws=None,
    grid_kws=None
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

#### **data :**

```python
plt.figure(figsize=(8,6))
sns.pairplot(data=df)
plt.title("Pair Plot")
plt.show()
```

#### **hue :**
##### Color grouping by category.

```python
plt.figure(figsize=(8,6))
sns.pairplot(data=df,hue='Performance_Rating')
plt.title("Pair Plot")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.pairplot(data=df,hue='Gender')
plt.title("Pair Plot")
plt.show()
```

#### **vars :**
##### Select specific columns.
##### **Example :** vars=[
    'Experience',
    'Salary',

]

```python
plt.figure(figsize=(8,6))
sns.pairplot(data=df,hue='Gender',vars=['Age','Salary_LPA','Project_Count'])
plt.title("Pair Plot")
plt.show()
```

#### **plot_kws :**
##### Customize scatter plots.
##### **Example :** plot_kws={'s':100,'alpha':0.7}
##### **Common Values :**
##### s=100
##### alpha=0.8

```python
plt.figure(figsize=(8,6))
sns.pairplot(data=df,hue='Gender',vars=['Experience_Years','Salary_LPA','Project_Count'],plot_kws={'s':100,'alpha':0.7})
plt.title("Pair Plot")
plt.show()
```

#### **diag_kws :**
##### Customize diagonal plots.
##### **Example :**
##### diag_kind='hist',
##### diag_kws={'bins':15}

```python
plt.figure(figsize=(8,6))
sns.pairplot(data=df,hue='Gender',vars=['Experience_Years','Salary_LPA','Project_Count'],diag_kind='hist',diag_kws={'bins':15})
plt.title("Pair Plot")
plt.show()
```
